```python
from ultralytics import YOLO

# Load a model
model = YOLO('yolo.pt')  # load a pretrained model (recommended for training)

# Use the model
model.to('cuda:0') # Move model to GPU
model.info()
```

```python
import os
import time
from PIL import Image
```

```python
%%time

dataset_path = f"{dataset.location}/test/images"
image_files = [os.path.join(dataset_path, f) for f in os.listdir(dataset_path) if f.endswith(('.jpg', '.jpeg', '.png'))]

print(f"Starting predictions on {len(image_files)} images...")

# Record start time
start_time = time.time()

for img_path in image_files:
    image = Image.open(img_path)
    # Perform prediction
    detections = model.predict(image)

# Record end time
end_time = time.time()

print("\nPrediction complete.")

# Calculate and display overall prediction time
total_prediction_time = end_time - start_time
print(f"Total Prediction Time for all images: {total_prediction_time:.4f} seconds")
```
